```python
import pandas as pd
```

```python
kanji_data = {}

for i in range(1, 6):
    url = f'https://jlptsensei.com/jlpt-n{i}-kanji-list/'
    data = pd.read_html(url)
    # data_df = pd.DataFrame(data[0])
    # data_clean = data_df.dropna()
    kanji_data[f'N{i}'] = data[0]

for level, data in kanji_data.items():
    md_file = data.to_markdown(index=False)
    with open(f'{level}_Kanji.md', 'w', encoding='utf-8') as f:
        f.write(md_file)
    print(f'Success Save Kanji level {level}')
```

```python
kanji = []
for i in range(1, 6):
    NLevel = pd.read_html(f'https://jlptsensei.com/jlpt-n{i}-kanji-list/')
    n_df = pd.DataFrame(NLevel[0])
    n_list = n_df.iloc[:, 1:2].dropna()
    n_list = n_list.values.tolist()
    print(f'Kanji level N{i}: \n{n_list}')
    kanji.append(n_list)
```

```python
with open('OnlyKanjiN1_N5', 'w', encoding='utf-8') as f:
    for k in kanji:
        f.write(f'{k}, ')
```
